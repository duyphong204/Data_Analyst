```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# BƯỚC QUAN TRỌNG: Đọc file dữ liệu của bạn
# Thay 'du_lieu_nha_dat.csv' bằng đường dẫn file thực tế trên máy bạn
df = pd.read_csv('VN_House_price_Clean2.csv') 

# Tạo figure với 2 biểu đồ con (subplots)
fig, axes = plt.subplots(1, 2, figsize=(16, 7))

# 1. Vẽ Boxplot cho Pháp lý (Legal status)
sns.boxplot(ax=axes[0], x='Legal status', y='Price', data=df, palette='Set2')
axes[0].set_title('Phân bố Giá theo Tình trạng Pháp lý', fontsize=14)
axes[0].tick_params(axis='x', rotation=30)
axes[0].set_xlabel('Tình trạng pháp lý')
axes[0].set_ylabel('Giá nhà (Tỷ VNĐ)')

# 2. Vẽ Boxplot cho Nội thất (Furniture state)
sns.boxplot(ax=axes[1], x='Furniture state', y='Price', data=df, palette='Set3')
axes[1].set_title('Phân bố Giá theo Tình trạng Nội thất', fontsize=14)
axes[1].tick_params(axis='x', rotation=30)
axes[1].set_xlabel('Mức độ nội thất')
axes[1].set_ylabel('Giá nhà (Tỷ VNĐ)')

plt.tight_layout()
plt.show()
```
