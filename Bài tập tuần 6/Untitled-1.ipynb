```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, FancyArrowPatch
from matplotlib.text import Text

def draw_class_diagram():
    fig, ax = plt.subplots(figsize=(10, 8))
    
    # Define class boxes
    classes = [
        {"name": "Student", "attributes": ["-name: String", "-id: String"], "methods": ["+getDetails(): String"], "x": 1, "y": 5},
        {"name": "Course", "attributes": ["-courseCode: String", "-title: String"], "methods": ["+getCourseInfo(): String"], "x": 5, "y": 5},
        {"name": "Enrollment", "attributes": ["-student: Student", "-course: Course", "-date: Date"], "methods": ["+enroll(): void"], "x": 3, "y": 2}
    ]
    
    # Draw class boxes
    for cls in classes:
        # Draw rectangle for class
        ax.add_patch(Rectangle((cls["x"], cls["y"]), 2, 2, fill=False))
        # Class name
        ax.text(cls["x"] + 1, cls["y"] + 1.8, cls["name"], ha="center", va="center", weight="bold")
        # Attributes
        attr_text = "\n".join(cls["attributes"])
        ax.text(cls["x"] + 1, cls["y"] + 1.4, attr_text, ha="center", va="center", fontsize=8)
        # Methods
        method_text = "\n".join(cls["methods"])
        ax.text(cls["x"] + 1, cls["y"] + 0.8, method_text, ha="center", va="center", fontsize=8)
    
    # Draw associations (arrows)
    # Student -> Enrollment
    ax.add_patch(FancyArrowPatch((3, 5), (3, 4), connectionstyle="arc3,rad=0.3", arrowstyle="-|>", color="black"))
    ax.text(3.2, 4.5, "enrolls", fontsize=8)
    
    # Course -> Enrollment
    ax.add_patch(FancyArrowPatch((5, 5), (4, 4), connectionstyle="arc3,rad=-0.3", arrowstyle="-|>", color="black"))
    ax.text(4.8, 4.5, "enrolled in", fontsize=8)
    
    # Set limits and remove axes
    ax.set_xlim(0, 8)
    ax.set_ylim(0, 8)
    ax.axis("off")
    
    plt.title("Class Diagram: Student Management System")
    plt.show()

if __name__ == "__main__":
    draw_class_diagram()
```
